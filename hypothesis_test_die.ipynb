```python
import numpy as np

# Observed data
n_rolls       = 60
observed_3s   = 19          # 3 came up 19 times in 60 rolls
expected_3s   = n_rolls / 6 # 10.0 under a fair die

print(f"Observed 3s : {observed_3s}")
print(f"Expected 3s : {expected_3s}")
```

## (b) Computing the p-value by simulation

```python
np.random.seed(42)
n_simulations = 100_000

# Simulate rolling a fair die 60 times, n_simulations times
rolls = np.random.randint(1, 7, size=(n_simulations, n_rolls))

# Test statistic: count of 3s in each simulated experiment
sim_counts = (rolls == 3).sum(axis=1)

# p-value: fraction of simulations where count of 3s >= observed
p_value = (sim_counts >= observed_3s).mean()

print(f"p-value = {p_value:.4f}")
```

## Exact p-value (Binomial)

We can also compute the exact p-value analytically,  
since under H₀ the count of 3s follows **Binomial(n=60, p=1/6)**.

```python
from scipy.stats import binom

# P(X >= 19) = 1 - P(X <= 18)
exact_p = 1 - binom.cdf(observed_3s - 1, n=n_rolls, p=1/6)

print(f"Exact p-value (binomial) = {exact_p:.4f}")
```

## Interpretation
With a p-value of approximately 0.003, we reject H₀ at the 5% significance level. The probability of observing 19 or more 3s in 60 rolls of a fair die is sufficiently small to conclude that the die is biased toward 3. The directional test statistic yields a smaller p-value than the non-directional alternative sum(abs(observed - expected)), reflecting the greater specificity of H₁.
